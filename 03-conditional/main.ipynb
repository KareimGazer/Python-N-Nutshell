## Control Flow

Thus far, every program we have written has been executed line by line in order. Instead of executing every line of code every single time a program is run, it is often useful to create sections of the program which are only executed in certain situations

control flow is the way in which a program executes code based on different conditions.
it uses logical expressions to determine whether a certain block of code should be executed or not.

it uses the keywords `if`, `elif`, and `else` to determine whether a certain block of code should be executed or not.
A conditional statement is one that that tells Python to run some code only when a certain condition is true.

```python
# the value given as input affects which parts of the program are executed.
# The program contains a conditional statement with a block of code which is executed only if the condition in the statement is true.

age = int(input("How old are you? "))

if age > 17: # if block
    print(f"You are of age {age}!")
    print("Here's a copy of GTA6 for you.")

print("Next customer, please!")
```

## Indentation
Python recognises that a block of code is part of a conditional statement if each line of code in the block is indented the same. That is, there should be a bit of whitespace at the beginning of every line of code within the code block. Each line should have the same amount of whitespace.

```python
password = input("Please type in a password: ")

if password == "kittycat":
    print("You knew the password!")
    print("You must be either the intended user...")
    print("...or quite an accomplished hacker.")

print("The program has finished its execution. Thanks and bye!")
```

## Combining conditions

### Logical operators

You can combine conditions with the logical operators `and` and `or`. The operator `and` specifies that all the given conditions must be true at the same time. The operator `or` specifies that at least one of the given conditions must be true.

```python
number = int(input("Please type in a number: "))
if number >= 5 and number <= 8:
    print("The number is between 5 and 8")
```

```python
number = int(input("Please type in a number: "))
if number < 5 or number > 8:
    print("The number is not within the range of 5 to 8")
```

### Boolean short circuting

when we are using the `or` operator in a conditional statement
- if the first expression evaluates to true, the second expression is not evaluated and `True` is returned
- if the first expression evaluates to false, the second expression is evaluated and the result is returned

this is known as short circuting. the opposite behaviour happens with `and` operator
- if the first expression evaluates to true, the second expression is evaluated and the result is returned
- if the first expression evaluates to false, the second expression is not evaluated and `False` is returned

```python
print(True or 5)
print(False or 5)
```

```python
print(True and 5)
print(False and 5)
```

# modulo
the modulo operator % is used to get the remainder of a division

```python
# 1, 5 is divided by 2 and the remainder is 1
# 5 = 2 * 2 + 1
print(5 % 2) # the remainder is one because it's an odd number
```

```python
# 2, 6 is divided by 2 and the remainder is 0
# 6 = 2 * 3
print(6 % 2) # the remainder is zero because it's an even number
```

so we can use the modulo operator to check if a number is even or odd

```python
number = 9
is_even = (number % 2 == 0)
if is_even:
    print("it's an even number")
else:
    print("it's an odd number")
```

```python
is_odd = (5 % 2) != 0
print(is_odd)
```

### Nested conditionals
Conditional statements can also be nested within other conditional statements. For example, the following program checks whether a number is above zero, and then whether it is odd or even:

```python
number = int(input("Please type in a number: "))

if number > 0:
    if number % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is negative or zero")
```

The same result can often be achieved using either nested conditional statements or conditions combined with logical operators. The example below is functionally no different from the example above, in the sense that it will print out the exactly same things with the same inputs:

```python
number = int(input("Please type in a number: "))

if number > 0 and number % 2 == 0:
    print("The number is even")
elif number > 0 and number % 2 != 0:
    print("The number is odd")
else:
    print("The number is negative or zero")
```

## Prime Numbers

a prime number is a number that is divisible by 1 and itself

```python
# 2 is a prime number because it is divisible by 1 and 2 only
# remember that when a number1 is divisable by another number2
# then number1 = number2 * something
# and number1 % number2 = 0
print(2 % 1 == 0)
print(2 % 2 == 0)
```

```python
# 3 is a prime number because it is divisible by 1, and 3 only
print(3 % 1 == 0)
print((3 % 2) == 0)
print(3 % 3 == 0)
```

```python
number = 6
divider = 2

is_prime = (number % divider != 0)
if is_prime:
    print("it's prime")
else:
    print("not prime")
```

```python
import math

def is_even(number):
    return number % 2 == 0

def is_prime(number):
    if number <= 1:
        return False
    # main logic
    for j in range(2, int(math.sqrt(number)) + 1):
        if number % j == 0:
            return False
    return True

number = int(input("Please, Enter a number: "))

if is_even(number):
    print(f"{number} is even.")
elif is_prime(number):
    print(f"{number} is prime.")
else:
    print(f"{number} is odd.")
```

An expression is a piece of code that resolves to some value like `5 + 9 * 2`. Within an expression, there are operators and operands.

Operators like `+` & `*` indicate what operation is to be performed (such as addition or subtraction).

Operands like `5`, `9`, and `2` are the numbers (or other values) on which we want to perform the operations.

### Expressions

there are many kinds of expressions in Python, and you can use them to perform a wide range of operations.

For example, the expression `5 + 9 * 2` is a mathematical expression that resolves to the value `25`, which is the result of adding `5` to `9` multiplied by `2`.

there is also logical expressions in Python, and they are used to resolve to boolean values like `True` or `False` that determine whether a condition is met or not.
For example, the expression `5 < 9` is a logical expression that resolves to `True` because `5` is less than `9`.
