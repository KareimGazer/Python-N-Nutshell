# Functions

Functions encapsulate blocks of code that execute when called. They serve as reusable units of code that perform specific tasks upon invocation.

A block of code within a function doesn't execute until explicitly invoked by calling that function. This control allows for efficient code organization and execution flow management.

```python
def sum_two_numbers(x, y):
    result = x + y
    return result
    print(x, y)

number = sum_two_numbers(5, 3)
print(number)
```

Arguments are values passed to a function when it's called. They enable customization and flexibility within functions by providing specific data or inputs required for the function's execution and behavior.

```python
def print_name(name):
    message = "Hello, " + name + "!"
    print(message)
    return "I'm happy to meet you!"

greetings = print_name("John")
print(greetings)
```

```python
print_name("John")
```

## Return Statements

A return statement is used to signify that a function has completed its task and is ready to provide a result or output back to the calling code.
It halts the function's execution and sends a value, if specified, back to the caller.

```python
def circle_area(radius, constant):
    pi = 3.14
    area = pi * (radius ** 2)
    return area

print(circle_area(5, 17.8) + circle_area(3))
```

```python
def calculation(num1, num2):
    result = num1 * num2
    return result

x = calculation(3, 4)
print(calculation(5, 2))
print(x)
```

## Types of Return Values

Python functions can return various data types such as integers, strings, lists, dictionaries, booleans, or even custom objects.

This versatility allows functions to convey different types of information or processed results to the calling code, enhancing the flexibility and usefulness of functions.

```python
6 % 2
```

```python
def some_func(number):
    if (number % 2) == 0:
        return number * 2
    else:
        return None

print(some_func(8))
```

## Working with strings

### String operations
Strings can be combined, or concatenated, with the + operator:

```python
begin = "ex"
end = "ample"
word = begin+end
print(word)
```

```python
word = "banana"
print(word*3)
```

Using string operations together with a loop we can write a program which draws a pyramid:

```python
n = 10 # number of layers in the pyramid
row = "*"

while n > 0:
    print(" " * n + row)
    row += "**"
    n -= 1
```

### Substrings and slices

```python
input_string = "presumptious"

print(input_string[0:3])
print(input_string[4:10])

# if the beginning index is left out, it defaults to 0
print(input_string[:3])

# if the end index is left out, it defaults to the length of the string
print(input_string[4:])
```

```python
# Searching for substrings - true or false

input_string = "test"

print("t" in input_string)
print("x" in input_string)
print("es" in input_string)
print("ets" in input_string)
```

```python
# Searching for substrings - index or -1

input_string = "test"

print(input_string.find("t"))
print(input_string.find("x"))
print(input_string.find("es"))
print(input_string.find("ets"))
```
